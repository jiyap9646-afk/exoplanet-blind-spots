```python
import pandas as pd
df = pd.read_csv("../data/Exoplanets.csv", comment="#")
print(df.shape)
print(list(df.columns))
df.head()
```

```python
cols = ["pl_name", "hostname", "discoverymethod", "disc_year",
        "pl_orbper", "pl_rade", "pl_bmasse", "pl_orbsmax",
        "pl_eqt", "st_teff", "st_rad", "sy_dist"]

df = df[cols]
print(df.shape)   
```

```python
df.isnull().sum()
```

```python
df["discoverymethod"].value_counts()
```

```python
df["discoverymethod"].value_counts(normalize=True)
```

```python
df.groupby("discoverymethod")[["pl_rade","pl_orbper"]].median()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.scatter(df["pl_orbper"], df["pl_rade"], s=5, alpha=0.4)
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Orbital period (days)")
plt.ylabel("Planet radius (Earth radii)")
plt.show()
```

```python
transit = df[df["discoverymethod"] == "Transit"]
other = df[df["discoverymethod"] != "Transit"]

plt.figure(figsize=(8, 5))
plt.scatter(transit["pl_orbper"], transit["pl_rade"], s=5, alpha=0.3, label="Transit")
plt.scatter(other["pl_orbper"], other["pl_rade"], s=8, alpha=0.6, label="All other methods")
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Orbital period (days)")
plt.ylabel("Planet radius (Earth radii)")
plt.legend()
plt.show()
```
